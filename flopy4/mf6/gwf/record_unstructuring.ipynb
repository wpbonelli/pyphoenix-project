```python
import attrs
import cattrs
from flopy4.mf6.gwf.oc import Oc
```

# Old `Oc.Steps`

## With `attrs.astuple`

```python
attrs.astuple(Oc.SaveRecord("head", Oc.Steps(all=True)))
```

```python
attrs.astuple(Oc.SaveRecord("head", Oc.Steps(steps=[1])))
```

## With `cattrs.unstructure`

```python
cattrs.unstructure(Oc.SaveRecord("head", Oc.Steps([1])))
```

```python
c = cattrs.Converter(unstruct_strat=cattrs.UnstructureStrategy.AS_TUPLE)
c.unstructure(Oc.SaveRecord("head", Oc.Steps(steps=[1])))
```

```python
c.register_unstructure_hook(Oc.Steps, lambda s: tuple([t for t in attrs.astuple(s) if t]))
c.unstructure(Oc.SaveRecord("head", Oc.Steps(steps=[1])))
```

# New record types

## With `attrs.astuple`

```python
from flopy4.mf6.gwf.oc import First
attrs.astuple(Oc.SaveRecord("head", First))
```

```python
from flopy4.mf6.gwf.oc import Steps
attrs.astuple(Oc.SaveRecord("head", Steps([1])))
```

```python
cattrs.unstructure(Oc.SaveRecord("head", Steps([1])))
```

```python
import cattrs
c = cattrs.Converter(unstructure_strat=cattrs.UnstructureStrategy.AS_TUPLE)
c.register_unstructure_hook(Oc.Steps, lambda s: tuple([t for t in attrs.astuple(s) if t]))
c.unstructure(Oc.SaveRecord("head", Oc.Steps([1])))
```
